```python
def add(a, b):
    return a + b

result = add(10, 20)
print("Sum:", result)
```

```python
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(7))
print(check_even_odd(12))
```

```python
def square(num):
    return num * num

print("Square:", square(5))
```

```python
def largest(a, b):
    if a > b:
        return a
    else:
        return b

print("Largest number:", largest(15, 25))
```

```python
def factorial(num):
    result = 1

    for i in range(1, num + 1):
        result = result * i

    return result

print("Factorial:", factorial(5))
```

```python
def check_palindrome(text):
    reversed_text = ""

    for char in text:
        reversed_text = char + reversed_text

    if text == reversed_text:
        return "Palindrome"
    else:
        return "Not a palindrome"

print(check_palindrome("madam"))
print(check_palindrome("hello"))
```

```python
def list_sum(numbers):
    total = 0

    for num in numbers:
        total = total + num

    return total

print("Total:", list_sum([10, 20, 30, 40]))
```

```python
def count_vowels(text):
    count = 0

    for char in text.lower():
        if char in "aeiou":
            count = count + 1

    return count

print("Vowel count:", count_vowels("Hello World"))
```

```python
def find_max(numbers):
    maximum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum

print("Maximum:", find_max([12, 45, 23, 67, 34]))
```

```python
def calculate_average(numbers):
    if len(numbers) == 0:
        return "List is empty"

    total = sum(numbers)
    average = total / len(numbers)

    return average

print("Average:", calculate_average([10, 20, 30, 40, 50]))
```
