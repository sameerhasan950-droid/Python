# 3. np.random.randn()
# ============================================================

# Purpose:
# Generates values from the standard normal distribution.
#
# Population Mean = 0
# Population Standard Deviation = 1
# Population Variance = 1
#

#

print("Variance:", x.var())

print("Standard Deviation:", x.std())

# Example 5: Simulate employee salaries

# Normal distributions are unbounded.

# ============================================================
# 7. np.ravel()
# ============================================================

# Purpose:
# Converts a multidimensional array into a 1D array.
#
# Syntax:
# np.ravel(array)
#
# It returns a view when possible.
# Otherwise, it returns a copy.

# Example 1: Convert 2D to 1D
x = np.array([[1, 2, 3],
              [4, 5, 6]])

print(np.ravel(x))

# Output:
# [1 2 3 4 5 6]

# Example 2: Using array method
print(x.ravel())

# Example 3: Convert 3D to 1D
data = np.random.randint(0, 100, (3, 3, 4))

print("Original Shape:", data.shape)

flattened = data.ravel()

print("Flattened Shape:", flattened.shape)
print(flattened)

# Example 4: Compare flatten() and ravel()
x = np.arange(6).reshape(2, 3)

a = x.ravel()
b = x.flatten()

print("Ravel:", a)
print("Flatten:", b)

# Difference:
# ravel() returns a view when possible.
# flatten() always returns a copy.
#
# A view may share memory with the original array.
# Modifying the view can modify the original array.

# Example 5: Demonstrate memory sharing
x = np.arange(6).reshape(2, 3)

a = x.ravel()
a[0] = 100

print("Original after ravel modification:")
print(x)

# Real-time applications:
# 1. Flattening image arrays
# 2. Preparing machine learning inputs
# 3. Converting multidimensional data
# 4. Feature preprocessing


# ============================================================
# 8. np.reshape()
# ============================================================

# Purpose:
# Changes the dimensions of an array
# without changing the number of elements.
#
# Syntax:
# np.reshape(array, new_shape)
#
# Important:
# Total elements before and after reshaping must match.

# Example 1: Convert 1D to 2D
data = np.arange(10)

print(data.reshape(1, 10))

# Output Shape:
# (1, 10)

# Example 2: Convert into 2 rows and 5 columns
print(data.reshape(2, 5))

# Example 3: Convert into 5 rows and 2 columns
print(data.reshape(5, 2))

# Example 4: Convert 12 values into a 3x4 matrix
x = np.arange(12)

print(x.reshape(3, 4))

# Example 5: Convert 24 values into 3D
x = np.arange(24)

print(x.reshape(2, 3, 4))

# Example 6: Automatically calculate one dimension
x = np.arange(12)

print(x.reshape(3, -1))

# Output Shape:
# (3, 4)

# Example 7: Convert a matrix back to 1D
x = np.arange(12).reshape(3, 4)

print(x.reshape(-1))

# 1. Preparing datasets for machine learning
# 2. Converting image dimensions
# 3. Transforming tabular data
# 4. Preparing neural network inputs

# Trigonometric functions operate on angles.
#
# NumPy expects angles in radians.
#
# 180 degrees = pi radians

# ------------------------------------------------------------
# 10.1 np.sin()
# ------------------------------------------------------------

# Purpose:
# Calculates sine of an angle in radians.

print(np.sin(1))

# Example: Sine of 30 degrees

print(np.sin(angle))

# Expected:
# 0.5 approximately


# ------------------------------------------------------------
# 10.2 np.cos()
# ------------------------------------------------------------

# Calculates cosine of an angle in radians.

print(np.cos(angle))

# Expected:

# ------------------------------------------------------------
# 10.3 np.tan()
# ------------------------------------------------------------

# Purpose:
# Calculates tangent of an angle in radians.
